### 1. Importación de la clase ManejadorCSV

En esta sección se importa la clase `ManejadorCSV` desde su archivo Python. También se configura la ruta de acceso a la carpeta de clases para permitir que Jupyter reconozca e importe correctamente el módulo.

```python
import sys
from pathlib import Path

# Agregar la carpeta de clases a la ruta de importación.
sys.path.append(str(Path("../clases").resolve()))

from manejador_csv import ManejadorCSV

print("Clase ManejadorCSV importada correctamente.")
```

### 2. Importación y visualización del archivo CSV

Se inicializa un objeto de la clase `ManejadorCSV` y se utiliza su método `importar_csv()` para leer el archivo original mediante `csv.reader`. Posteriormente, se muestran los encabezados, la cantidad de registros y los primeros cinco datos para conocer su estructura.

```python

from pathlib import Path

# Definir la ruta del archivo CSV original.
ruta_csv = Path("../datos/student_performance_original.csv")

# Crear el objeto de la clase ManejadorCSV.
csv_manager = ManejadorCSV()

# Importar los datos.
encabezados_csv, datos_csv = csv_manager.importar_csv(ruta_csv)

# Mostrar los resultados.
print("Encabezados del CSV:")
print(encabezados_csv)

print("\nCantidad de registros:", len(datos_csv))

print("\nPrimeros 5 registros:")
for registro in datos_csv[:5]:
    print(registro)
```

### 3. Modificación y almacenamiento del archivo CSV

Se modifica la puntuación de lectura del primer estudiante, cambiando su valor original a 95 mediante el método `modificar_valor()`. Finalmente, se utiliza `guardar_csv()` para almacenar los datos modificados en un nuevo archivo, conservando intacto el CSV original.

```python

# Modificar la puntuación de lectura del primer estudiante.
valor_anterior, valor_nuevo = csv_manager.modificar_valor(
    numero_fila=0,
    columna="reading_score",
    nuevo_valor=95
)

print(f"Puntuación anterior: {valor_anterior}")
print(f"Puntuación nueva: {valor_nuevo}")

# Guardar los datos modificados en un archivo nuevo.
ruta_csv_modificado = Path("../datos/student_performance_modificado.csv")

csv_manager.guardar_csv(ruta_csv_modificado)

print(f"\nArchivo guardado en: {ruta_csv_modificado}")
```

### 4. Importación y visualización del archivo JSON

En esta sección se importa la clase `ManejadorJSON` y se utiliza para cargar el archivo JSON original mediante el método `importar_json()`.

Se identifica el tipo de estructura de los datos y se muestra un registro de ejemplo para conocer sus campos y contenido.

```python
from manejador_json import ManejadorJSON

# Definir la ruta del archivo JSON original.
ruta_json = Path("../datos/math23k_original.json")

# Crear el objeto de la clase ManejadorJSON.
json_manager = ManejadorJSON()

# Importar los datos.
datos_json = json_manager.importar_json(ruta_json)

# Mostrar la estructura y un registro de ejemplo.
print("Tipo de estructura:", type(datos_json).__name__)
print("Cantidad de registros:", json_manager.registros_procesados)

print("\nPrimer registro:")
if isinstance(datos_json, list):
    print(datos_json[0])
elif isinstance(datos_json, dict):
    print(next(iter(datos_json.items())))
else:
    print(datos_json)
```

### 5. Modificación y almacenamiento del archivo JSON

En esta sección se modifica el campo `system` del primer registro utilizando el método `modificar_registro()`. Posteriormente, se utiliza `guardar_json()` para almacenar los datos actualizados en un nuevo archivo, conservando intacto el JSON original.

```python
# Modificar el campo system del primer registro.
valor_anterior, valor_nuevo = json_manager.modificar_registro(
    indice=0,
    clave="system",
    nuevo_valor="Resuelve el problema matemático paso a paso."
)

print("Valor anterior:", repr(valor_anterior))
print("Valor nuevo:", valor_nuevo)

# Guardar los datos modificados en un archivo nuevo.
ruta_json_modificado = Path("../datos/math23k_modificado.json")

json_manager.guardar_json(ruta_json_modificado)

print("\nArchivo guardado en:", ruta_json_modificado)
```

### 6. Creación de la clase GeneradorResumen

En esta sección se implementa la clase `GeneradorResumen`, encargada de organizar la información del procesamiento de los archivos CSV y JSON. Mediante sus métodos se agregan secciones descriptivas y se genera un archivo de texto (`resumen.txt`) con los resultados de la actividad.

```python
from generador_resumen import GeneradorResumen

# Crear el objeto encargado de generar el resumen.
resumen = GeneradorResumen()

# Registrar las fuentes de los archivos originales.
resumen.agregar_seccion(
    "FUENTES DE LOS DATOS",
    "CSV: https://huggingface.co/datasets/mstz/student_performance/blob/main/math/train.csv\n"
    "JSON: https://huggingface.co/datasets/hehuanhao/fusion-dataset/blob/main/math23k/train.json"
)

# Registrar los archivos importados.
resumen.agregar_seccion(
    "ARCHIVOS IMPORTADOS",
    "student_performance_original.csv\n"
    "math23k_original.json"
)

# Registrar las modificaciones realizadas.
resumen.agregar_seccion(
    "MODIFICACIONES REALIZADAS",
    "CSV: Se modificó reading_score del primer registro de 70 a 95.\n"
    "JSON: Se modificó el campo system del primer registro, que estaba vacío."
)

# Registrar las cantidades de registros.
resumen.agregar_seccion(
    "CANTIDAD DE REGISTROS",
    f"CSV: {csv_manager.registros_procesados}\n"
    f"JSON: {json_manager.registros_procesados}"
)

# Registrar las clases y métodos utilizados.
resumen.agregar_seccion(
    "CLASES Y METODOS UTILIZADOS",
    "ManejadorCSV: importar_csv(), modificar_valor(), guardar_csv().\n"
    "ManejadorJSON: importar_json(), modificar_registro(), guardar_json().\n"
    "GeneradorResumen: agregar_seccion(), generar_resumen()."
)

# Registrar las rutas de los archivos generados.
resumen.agregar_seccion(
    "ARCHIVOS GENERADOS",
    "datos/student_performance_modificado.csv\n"
    "datos/math23k_modificado.json\n"
    "reportes/resumen.txt"
)

# Generar el archivo de resumen.
ruta_resumen = Path("../reportes/resumen.txt")
resumen.generar_resumen(ruta_resumen)

print("Resumen generado correctamente en:", ruta_resumen)
```

```python
# Leer y mostrar el contenido del resumen generado.
with open(ruta_resumen, "r", encoding="utf-8") as archivo:
    contenido_resumen = archivo.read()

print(contenido_resumen)
```
